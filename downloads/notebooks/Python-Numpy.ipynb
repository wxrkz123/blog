## 五、索引与切片

一维数组的索引和切片与列表相同。二维数组用逗号分隔行和列：

```python
m = np.arange(1, 13).reshape(3, 4)
print(m)
print(m[1, 2])        # 第 2 行第 3 列（下标从 0 开始）
print(m[0])           # 第 1 行
print(m[:, 1])        # 第 2 列
print(m[:2, 1:3])     # 前 2 行、第 2–3 列
```

**布尔索引**是 NumPy 最常用的功能之一：用一个条件筛选元素。

```python
returns = np.array([0.012, -0.008, 0.021, -0.015, 0.004])
print(returns > 0)                  # 每个元素是否大于 0
print(returns[returns > 0])         # 只取正收益
print((returns > 0).sum())          # 上涨的天数：True 当作 1，False 当作 0
```

**切片是视图，不是副本。** 这是和列表最大的不同：对切片的修改会影响原数组。

```python
x = np.arange(5)
view = x[1:4]
view[0] = 100
print(x)                  # 原数组也被修改了

y = np.arange(5)
copy = y[1:4].copy()      # 需要独立的副本时，显式调用 copy()
copy[0] = 100
print(y)
```

## 八、向量化运算与广播

数组之间的加减乘除，都是**逐元素**进行的，不需要写循环：

```python
prices = np.array([10.0, 10.5, 10.2, 10.8, 11.0])
daily = prices[1:] / prices[:-1] - 1          # 日收益率：今天的价格 / 昨天的价格 - 1
print(daily.round(4))
print(np.diff(prices))                        # 相邻两项的差
print(prices * 100)                           # 每个元素乘以 100（买 100 股的市值）
print(prices ** 2)
```

`prices * 100` 是一个数组和一个数相乘，NumPy 会把 100 “扩展”成和数组一样的形状，再逐元素相乘。这种机制叫**广播**（broadcasting），它也适用于形状不同的两个数组：

```python
holdings = np.array([[100], [200], [300]])    # 3 个账户各持有多少股，形状 (3, 1)
print(holdings * prices)                      # (3, 1) 和 (5,) 广播成 (3, 5)：每个账户每天的市值
```

矩阵乘法用 `@` 运算符（或 `np.dot`）：

```python
weights = np.array([0.5, 0.3, 0.2])             # 组合中 3 只股票的权重
asset_returns = np.array([0.10, 0.06, -0.02])   # 3 只股票的收益率
print(f"组合收益率：{weights @ asset_returns:.3f}")
```

## 九、统计函数

```python
r = np.array([0.012, -0.008, 0.021, -0.015, 0.004, 0.009, -0.003])
print(f"平均日收益：{r.mean():.4f}")
print(f"标准差（总体）：{r.std():.4f}")
print(f"标准差（样本，ddof=1）：{r.std(ddof=1):.4f}")
print(f"最大 / 最小：{r.max()} / {r.min()}，最大值在第 {r.argmax()} 天")
print(f"累计收益：{np.prod(1 + r) - 1:.4f}")
print("净值：", np.cumprod(1 + r).round(4))
```

二维数组可以按行或按列统计。下面是 3 只股票 4 天的收益率，每一列是一只股票：

```python
R = np.array([[ 0.01,  0.02, -0.01],
              [-0.02,  0.01,  0.00],
              [ 0.03,  0.02,  0.01],
              [ 0.00, -0.01,  0.02]])
print("每只股票的平均收益：", R.mean(axis=0))
print("每天的平均收益：", R.mean(axis=1))
print("相关系数矩阵：")
print(np.corrcoef(R, rowvar=False).round(2))    # rowvar=False：每一列是一个变量
```

## 十一、随机数与蒙特卡洛模拟

NumPy 推荐用 `np.random.default_rng` 创建一个随机数生成器：

```python
rng = np.random.default_rng(42)              # 42 是种子
print(rng.normal(0, 1, 5).round(3))          # 正态分布：均值、标准差、个数
print(rng.integers(1, 7, 10))                # 1 到 6 的随机整数（掷骰子）
print(rng.uniform(0, 1, 3).round(3))         # [0, 1) 均匀分布
```

抽样的个数越多，样本的统计量越接近理论值：

```python
for n in [100, 10_000, 1_000_000]:
    z = rng.standard_normal(n)
    print(f"抽样 {n:>9,} 次：均值 {z.mean():+.4f}，标准差 {z.std():.4f}")

t5 = rng.standard_t(5, 1_000_000)
print(f"\nt 分布（自由度 5）：标准差 {t5.std():.3f}（理论值 {np.sqrt(5 / 3):.3f}），"
      f"超过 4 的比例 {(np.abs(t5) > 4).mean():.4%}")
print(f"标准正态分布：超过 4 的比例 {(np.abs(rng.standard_normal(1_000_000)) > 4).mean():.4%}")
```

**蒙特卡洛模拟**：用随机数模拟许多种可能的未来。假设一只股票的年化收益率为 8%、年化波动率为 20%，每天的收益率服从正态分布，模拟一年后 100 万元的投资会变成多少：

```python
n_paths, n_days = 100_000, 243
mu, sigma = 0.08 / n_days, 0.20 / np.sqrt(n_days)
daily_returns = rng.normal(mu, sigma, size=(n_paths, n_days))   # 10 万条路径 × 243 天，一次生成
final = 1_000_000 * np.prod(1 + daily_returns, axis=1)
print(f"一年后的平均值：{final.mean():,.0f} 元")
print(f"中位数：{np.median(final):,.0f} 元")
print(f"有 5% 的可能低于：{np.percentile(final, 5):,.0f} 元")
print(f"亏损的概率：{(final < 1_000_000).mean():.1%}")
```

### 现值

函数是 `npf.pv(rate, nper, pmt, fv)`，最后一个参数是**期末的一次性金额**（不是现值）：

```python
# 一份年金：未来 20 年每年年末领 5 万元，折现率 4%
print(f"这份年金今天值：{npf.pv(0.04, 20, 50_000, 0):,.2f} 元")
# 一张债券：面值 100 元，票面利率 5%（每年付息 5 元），5 年到期，市场利率 4%
print(f"债券价格：{npf.pv(0.04, 5, 5, 100):,.2f} 元")
```

### 净现值与内部收益率

使 NPV 等于 0 的折现率叫**内部收益率**（IRR）。

```python
# 期初投资 100 万元，之后 5 年每年年末分别收回 20、30、35、30、20 万元
cash_flows = np.array([-100, 20, 30, 35, 30, 20]) * 10_000
print(f"折现率 8% 时的净现值：{npf.npv(0.08, cash_flows):,.2f} 元")
print(f"内部收益率：{npf.irr(cash_flows):.2%}")
print(f"验证：用 IRR 折现，净现值为 {npf.npv(npf.irr(cash_flows), cash_flows):.6f}")
```

### 等额本息房贷

贷款 100 万元，年利率 3.6%，30 年，按月等额本息还款。月利率是 $3.6\% / 12$，期数是 360：

```python
rate, nper, loan = 0.036 / 12, 30 * 12, 1_000_000
payment = npf.pmt(rate, nper, loan)
print(f"每月还款：{-payment:,.2f} 元，总利息：{-payment * nper - loan:,.2f} 元")

months = np.array([1, 2, 120, 240, 360])
principal = npf.ppmt(rate, months, nper, loan)      # 每月还款中的本金部分
interest = npf.ipmt(rate, months, nper, loan)       # 每月还款中的利息部分
for m, p, i in zip(months, principal, interest):
    print(f"第 {m:3d} 个月：本金 {-p:8,.2f}，利息 {-i:8,.2f}")
```
