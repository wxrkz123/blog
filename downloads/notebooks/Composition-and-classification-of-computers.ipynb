## 三、内存：能同时处理多少数据

一个 DataFrame 占多少内存，可以直接估算：每个浮点数或整数占 8 个字节。100 万行、20 列的数值数据就是 $10^6 \times 20 \times 8 = 1.6 \times 10^8$ 字节，约 160 MB：

```python
import numpy as np
import pandas as pd

df = pd.DataFrame(np.zeros((1_000_000, 20)), columns=[f"x{i}" for i in range(20)])
print(f"数值列：{df.memory_usage(deep=True).sum() / 1e6:.0f} MB")

df["code"] = [f"{i:06d}" for i in range(len(df))]        # 加一列 6 位的股票代码（文本）
print(f"加一列文本（{df['code'].dtype}）：{df.memory_usage(deep=True).sum() / 1e6:.0f} MB")
as_object = df.assign(code=df["code"].astype(object))           # pandas 2 及以前的默认存法
print(f"同一列存成 Python 对象：{as_object.memory_usage(deep=True).sum() / 1e6:.0f} MB")
```

## 六、查看自己电脑的配置

几行 Python 就能查看主要的配置（输出因电脑而异）：

```python
import os
import platform
import shutil

print("操作系统：", platform.system(), platform.release())
print("处理器：  ", platform.processor() or platform.machine())
print("逻辑核心：", os.cpu_count())
total, used, free = shutil.disk_usage("/")
print(f"硬盘：     共 {total / 1e9:.0f} GB，剩余 {free / 1e9:.0f} GB")
```
